```python
corpus="""Hello Welcome,to Krish Naik's NLP Tutorials.
Please do watch the entire course! to become expert in NLP.
"""
```

```python
print(corpus)
```

```python
import nltk

nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
##  Tokenization
## paragraphs->sentence
from nltk.tokenize import sent_tokenize
```

```python
documents=sent_tokenize(corpus)
print("Tokenized Paragraphs:",documents)
```

```python
type(documents)
```

```python
for sentence in documents:
    print(sentence)
```

```python
## Tokenization 
## Paragraph-->words
from nltk.tokenize import word_tokenize

words=word_tokenize(corpus)
print("Tokenized Words:",words)
```

```python
type(words)
```

```python
## sentence--->words
for sentence in documents:
    words=word_tokenize(sentence)
    print(words)
```

```python
from nltk.tokenize import wordpunct_tokenize
```

```python
words=wordpunct_tokenize(corpus)
print("Tokenized Words using wordpunct_tokenize:",words)
```

```python
from nltk.tokenize import TreebankWordTokenizer

tokenizer = TreebankWordTokenizer()
words = tokenizer.tokenize(corpus)
print("Tokenized Words using TreebankWordTokenizer:", words)
```
